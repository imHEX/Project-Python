```python
class person:
    def __init__(self, name, age, id):
        self.name = name
        self.age = age
        self.id = id
    
    def grow_up(self):
        self.age += 1
```

```python
class employee(person):
    def __init__(self, name, age, id, job):
        super().__init__(name, age, id)
        self.job = job
    
    def work(self):
        print(f"{self.name} is workin")
```

```python
class manager(employee):
    def __init__(self, name, age, id, job, income):
        super().__init__(name, age, id, job)
        self.income = income
        
    def salary(self):
        self.income += 1000
```

```python
class stethoscope:
    def __init__(self, brand):
        self.brand = brand
        self.battery = 100
        
    def listen(self):
      if self.battery > 0:
            print("Listening to heart beat...")
            self.battery -= 1
      else:
            print("battery is dead please recharge")
```

```python
class Doctor:
    def __init__(self, name):
        self.name = name
        
    def check_patient(self, stethoscope):
        print(f"{self.name} is checking the patient with {stethoscope.brand}")
        stethoscope.listen()
```

```python
st = stethoscope('Littman')
```

```python
dr = Doctor("Dr.John")
```

```python
dr.check_patient(st)
```

```python
class player:
    def __init__(self, name):
        self.name = name
```

```python
class team:
    def __init__(self, name):
        self.name = name
        self.players = []
        
    def add_player(self, player):
        self.players.append(player)
        
    def remove_player(self , player):
        self.players.remove(player)
        
    def display_player(self):
        for player in self.players:
            print(player.name)
```

```python
p1 = player("Max")
p2 = player("Jeff")
p3 = player("Mike")
```

```python
t1 = team("Team A")
```

```python
t1.display_player()
```

```python
t1.add_player(p1)
t1.add_player(p2)
t1.add_player(p3)
```

```python
t1.display_player()
```

```python
class processor:
    def __init__(self, brand):
        self.brand = brand
    def process(self):
        print("Processing...")

class ram:
    def __init__(self, brand, size):
        self.brand = brand
        self.size = size
    def reading(self):
        print("Ram is reading The Data...")

class harddrive:
    def __init__(self, brand, size):
        self.brand = brand
        self.size = size
    def storing(self):
        print("Hard Drive is storing The Data...")
        
```

```python
class computer:
    def __init__(self, processor, ram, harddrive):
        self.processor = processor
        self.ram = ram
        self.harddrive = harddrive
        
    def boot(self):
        print("Booting up the PC...")
        self.processor.process()
        self.ram.reading()
        self.harddrive.storing()
        
```

```python
p = processor("AMD")
r = ram("Corsair", 16)
h = harddrive("WD", 2)

pc1 = computer(p, r, h)
```

```python
pc1.boot()
```
